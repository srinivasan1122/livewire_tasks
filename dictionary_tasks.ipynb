```python
#find highest salary
employees = {"Arun": 45000, "Priya": 52000, "Kumar": 48000, "Divya": 60000}

highest = 0
name = ""

for employee in employees:
    if employees[employee] > highest:
        highest = employees[employee]
        name = employee

print(name, "with salary", highest)
```

```python
#Sort Dictionary by Values
employees = {
    "Arun": 45000,
    "Priya": 52000,
    "Kumar": 48000,
    "Divya": 60000
}

result = list(employees.items())

for i in range(len(result)):
    for j in range(i + 1, len(result)):
        if result[i][1] > result[j][1]:
            temp = result[i]
            result[i] = result[j]
            result[j] = temp

print(result)
```

```python
#Sort Dictionary by Keys
employees = {
    "Priya": 102,
    "Arun": 101,
    "Kumar": 103
}

result = list(employees.items())

for i in range(len(result)):
    for j in range(i + 1, len(result)):
        if result[i][0] > result[j][0]:
            temp = result[i]
            result[i] = result[j]
            result[j] = temp

print(dict(result))
```

```python
#Word Frequency
sentence = "python java python sql java python"
words = sentence.split()
freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

for word in freq:
    print(word, ":", freq[word])
```

```python
#Character Frequency
word = "programming"
freq = {}

for char in word:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

for char in freq:
    print(char, ":", freq[char])
```

```python
#highest frequency element
arr = [10, 20, 10, 30, 20, 10]
freq = {}

for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

highest = 0
result = 0

for num in freq:
    if freq[num] > highest:
        highest = freq[num]
        result = num

print(result)
```

```python
#group employees by department
employees = {
    "Arun": "IT",
    "Priya": "HR",
    "Kumar": "IT",
    "Divya": "Finance",
    "Rahul": "IT"
}

result = {}

for name in employees:
    department = employees[name]

    if department not in result:
        result[department] = []

    result[department].append(name)

print(result)
```

```python
#product inventory
stock = {
    "Laptop": 15,
    "Mouse": 8,
    "Keyboard": 25,
    "Monitor": 5
}

highest = 0
lowest = 100000
highest_product = ""
lowest_product = ""
total = 0
below = []

for product in stock:
    if stock[product] > highest:
        highest = stock[product]
        highest_product = product

    if stock[product] < lowest:
        lowest = stock[product]
        lowest_product = product

    total += stock[product]

    if stock[product] < 10:
        below.append(product)

print("Highest:", highest_product)
print("Lowest:", lowest_product)
print("Total:", total)
print("Below 10:", below)
```

```python
#student ranking
students = {
    "Arun": 85,
    "Priya": 92,
    "Kumar": 78,
    "Divya": 95,
    "Rahul": 88
}

result = list(students.items())

for i in range(len(result)):
    for j in range(i + 1, len(result)):
        if result[i][1] < result[j][1]:
            temp = result[i]
            result[i] = result[j]
            result[j] = temp

print("Highest:", result[0])
print("Lowest:", result[-1])
print("Sorted:", result)
print("Top 3:", result[:3])
```
