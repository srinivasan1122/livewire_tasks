```python
arr = [101, 205, 304, 405, 502]
target = 304

for i in range(len(arr)):
    if arr[i] == target:
        print("Element found at index", i)
        break
```

```python
arr = [45, 12, 89, 67, 34, 90]

largest = second = float('-inf')

for num in arr:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num

print(second)
```

```python
arr = [45, 12, 89, 67, 34, 90]

smallest = second = float('inf')

for num in arr:
    if num < smallest:
        second = smallest
        smallest = num
    elif num < second and num != smallest:
        second = num

print(second)
```

```python
arr = [64, 34, 25, 12, 22, 11, 90]

for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] > arr[j]:
            arr[i], arr[j] = arr[j], arr[i]

print(arr)
```

```python
arr = [10, 20, 10, 30, 20, 40]
result = []

for num in arr:
    if num not in result:
        result.append(num)

print(result)
```

```python
arr = [1, 2, 3, 5, 6]
n = 6

total = n * (n + 1) // 2
missing = total - sum(arr)

print(missing)
```

```python
A = [10, 20, 30, 40]
B = [30, 40, 50, 60]
result = []

for num in A:
    if num in B and num not in result:
        result.append(num)

print(result)
```

```python
arr = [0, 5, 0, 3, 8, 0, 2]

result = []

for num in arr:
    if num != 0:
        result.append(num)

for num in arr:
    if num == 0:
        result.append(num)

print(result)
```

```python
arr = [10, 20, 10, 30, 20, 10]
freq = {}

for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

for num in freq:
    print(num, ":", freq[num])
```

```python
prices = [120.5, 132.0, 118.4, 145.2, 128.9]

print("Highest:", max(prices))
print("Lowest:", min(prices))
print("Average:", sum(prices) / len(prices))
```
