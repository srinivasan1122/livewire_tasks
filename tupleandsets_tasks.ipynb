```python
#tuple sorting
t = (50, 10, 40, 20, 30)
t2 = tuple(sorted(t))
print(t2)
```

```python
#Find Maximum and Minimum (without max() and min())
t = (15, 82, 24, 95, 33)
maximum = t[0]
minimum = t[0]

for n in t:
    if n > maximum:
        maximum = n
    if n < minimum:
        minimum = n

print("Max:", maximum)
print("Min:", minimum)
```

```python
#Count Duplicate Values
t = (10, 20, 10, 30, 20, 10)

for n in t:
    if t.index(n) == t.index(n):
        if t.index(n) == t.index(n):
            pass

for n in t:
    if t.index(n) == t.index(n):
        print(n, ":", t.count(n))
```

```python
#remove duplicate values
t = (1, 2, 2, 3, 4, 4, 5)
result = []

for number in t:
    if number not in result:
        result.append(number)

print(tuple(result))
```

```python
#tuple comparison
t1 = (1, 2, 3)
t2 = (3, 2, 1)

print("Equal Elements:", sorted(t1) == sorted(t2))
```

```python
#find common elements
a = (10, 20, 30, 40)
b = (30, 40, 50, 60)
result = ()

for n in a:
    if n in b:
        result = result + (n,)

print(result)
```

```python
#tuple of employee records
employees = (
    ("E101", "Arun", 45000),
    ("E102", "Priya", 52000),
    ("E103", "Kumar", 48000)
)

highest = employees[0]

for employee in employees:
    if employee[2] > highest[2]:
        highest = employee

print("Highest Salary:", highest)
```

```python
#sort employee records
employees = (
    ("E101", "Arun", 45000),
    ("E102", "Priya", 52000),
    ("E103", "Kumar", 48000)
)

result = list(employees)

for i in range(len(result)):
    for j in range(i + 1, len(result)):
        if result[i][2] > result[j][2]:
            temp = result[i]
            result[i] = result[j]
            result[j] = temp

print(tuple(result))
```

```python
#Find Duplicate Values
arr = [10, 20, 30, 10, 20, 40]

duplicates = set()

for num in arr:
    if arr.count(num) > 1:
        duplicates.add(num)

print("Duplicates:", duplicates)
```

```python
#subset validation
required = {"Python", "SQL", "Git"}
candidate = {"Python", "SQL", "Git", "Java"}

if required.issubset(candidate):
    print("Eligible")
else:
    print("Not Eligible")
```
