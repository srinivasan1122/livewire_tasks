```python
#5.	Write a Program to Solve Quadratic Equation [ (a+b)2 formula , (a-b)2 formula ,a2-b2  formula ]
a=2
b=4
print("(a+b)^2 is ",a**2+b**2+2*(a*b))
print("(a-b)^2 is ",a**2+b**2-2*(a*b))
print("(a^2-b^2) is ",(a+b)*(a-b))
```
