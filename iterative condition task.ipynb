```python
n = int(input())
i = 1
count = 0

while count < n:
    if i % 3 == 0 or i % 5 == 0 or i % 7 == 0:
        print(i, end=" ")
        count += 1
    i += 1

print()
```

```python
n = int(input())
prime = True

if n < 2:
    prime = False
else:
    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

if prime:
    print("Yes")
else:
    print("No")
```

```python
n = int(input())

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")

print()
```

```python
n = int(input())
even_sum = 0
odd_sum = 0

for i in range(n):
    num = int(input())

    if num % 2 == 0:
        even_sum += num
    else:
        odd_sum += num

print("Even sum:", even_sum)
print("Odd sum:", odd_sum)
```
