```python
#1.	Find the largest number among three integers.
a = 10
b = 25
c = 15

if a >= b and a >= c:
    print(a)
elif b >= a and b >= c:
    print(b)
else:
    print(c)
```

```python
#2.	Find the smallest number among three integers.
a = 10
b = 25
c = 15

if a <= b and a <= c:
    print(a)
elif b <= a and b <= c:
    print(b)
else:
    print(c)
```

```python
#3.	Check if a given number is greater than 0, if yes then print 'Positive'. If the given number is lesser than 0, then print 'Negative'. If the given number is exactly equal to 0, then print 'Zero'.
num = -5

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
#5.	Write a  program that functions as a basic calculator. The program will prompt the user to input two numbers and a mathematical operation (+, -, x, /). It will then perform the selected operation and display the result on the screen.
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operation = input("Enter operation (+, -, x, /): ")

if operation == "+":
    print(num1 + num2)

elif operation == "-":
    print(num1 - num2)

elif operation == "x":
    print(num1 * num2)

elif operation == "/":
    print(num1 / num2)

else:
    print("Invalid operation")
```

```python
#6.	Check whether the given number is a multiple of 5, 3, and 7.
num = int(input("Enter a number: "))

if num % 5 == 0 and num % 3 == 0 and num % 7 == 0:
    print("The number is a multiple of 5, 3, and 7")
else:
    print("The number is not a multiple of 5, 3, and 7")
```
